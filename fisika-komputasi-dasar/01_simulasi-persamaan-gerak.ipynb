```python
import sympy as sp

# Definisikan variabel waktu (t) sebagai simbol matematika
t = sp.Symbol('t')

# Masukkan persamaan percepatan a(t)
a = 4*t - 3

# Cari persamaan kecepatan v(t) dengan mengintegralkan a(t)
# Ditambahkan kecepatan awal v(0) = 2 m/s sebagai konstanta
v = sp.integrate(a, t) + 2

# Cari persamaan posisi x(t) dengan mengintegralkan v(t)
# Ditambahkan posisi awal x(0) = 0 m sebagai konstanta
x = sp.integrate(v, t) + 0

# Tampilkan hasil persamaan dengan format matematika yang rapi
print("Hasil Perhitungan Simbolik:")
print("-" * 27)
display(sp.Eq(sp.Function('v')(t), v))
display(sp.Eq(sp.Function('x')(t), x))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Rentang waktu (t) dari 0 sampai 10 detik
t = np.linspace(0, 10, 500)

# Persamaan fungsi matematika gerak benda
a = 4 * t - 3
v = 2 * (t**2) - 3 * t + 2
x = (2/3) * (t**3) - (1.5) * (t**2) + 2 * t

# Membuat dekorasi dan layout grafik (3 baris, 1 kolom)
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(5, 10))
fig.suptitle('Grafik Kinematika Gerak Benda (0 - 10 Detik)', fontsize=16, fontweight='bold')

# --- GRAFIK 1: POSISI TERHADAP WAKTU ---
ax1.plot(t, x, color='blue', linewidth=2.5, label='x(t)')
ax1.set_title('Posisi vs Waktu')
ax1.set_xlabel('Waktu (sekon)')
ax1.set_ylabel('Posisi (meter)')
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend()

# --- GRAFIK 2: KECEPATAN TERHADAP WAKTU ---
ax2.plot(t, v, color='green', linewidth=2.5, label='v(t)')
ax2.set_title('Kecepatan vs Waktu')
ax2.set_xlabel('Waktu (sekon)')
ax2.set_ylabel('Kecepatan (m/s)')
ax2.grid(True, linestyle='--', alpha=0.6)
ax2.legend()

# --- GRAFIK 3: PERCEPATAN TERHADAP WAKTU ---
ax3.plot(t, a, color='red', linewidth=2.5, label='a(t)')
ax3.set_title('Percepatan vs Waktu')
ax3.set_xlabel('Waktu (sekon)')
ax3.set_ylabel('Percepatan (m/s²)')
ax3.grid(True, linestyle='--', alpha=0.6)
ax3.legend()

# Menampilkan grafik di layar
plt.tight_layout()
plt.show()
```

```python
import sympy as sp

# Definisikan simbol t
t = sp.Symbol('t')

# Persamaan posisi s(t)
s = 4*t**3 + t**2 - 3

# Menghitung persamaan kecepatan v(t) - Turunan pertama
v = sp.diff(s, t)

# Menghitung persamaan percepatan a(t) - Turunan kedua
a = sp.diff(v, t)

# Tampilkan hasil persamaan dengan format matematika yang rapi
print("Hasil Perhitungan Simbolik:")
print("-" * 27)
display(sp.Eq(sp.Function('v')(t), v))
display(sp.Eq(sp.Function('a')(t), a))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Membuat data waktu (t) dari 0 hingga 2 detik
t = np.linspace(0, 2, 500)

# Mendefinisikan persamaan fisikanya
s = 4 * t**3 + t**2 - 3
v = 12 * t**2 + 2 * t
a = 24 * t + 2

# Membuat visualisasi grafik dalam satu plot
plt.figure(figsize=(10, 6))

# Memplot masing-masing kurva
plt.plot(t, s, label='Posisi $s(t) = 4t^3 + t^2 - 3$ (m)', color='red', linewidth=2)
plt.plot(t, v, label='Kecepatan $v(t) = 12t^2 + 2t$ (m/s)', color='blue', linewidth=2)
plt.plot(t, a, label='Percepatan $a(t) = 24t + 2$ ($m/s^2$)', color='green', linewidth=2)

# Menambahkan elemen pelengkap informasi
plt.title('Grafik Kinematika: Posisi, Kecepatan, dan Percepatan terhadap Waktu', fontsize=14)
plt.xlabel('Waktu $t$ (detik)', fontsize=12)
plt.ylabel('Nilai Skala (m, m/s, $m/s^2$)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)

# Menampilkan grafik di layar
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Membuat data waktu (t) dari 0 hingga 20 detik
t = np.linspace(0, 20, 500)

# Mendefinisikan persamaan fisikanya
x = 0.1 * t**3 - (1/300) * t**4
v = 0.3 * t**2 - (1/75) * t**3
a = 0.6 * t - 0.04 * t**2

# Membuat visualisasi grafik dalam satu plot
plt.figure(figsize=(10, 6))

# Memplot masing-masing kurva
plt.plot(t, x, label='Posisi $x(t)$ (m)', color='red', linewidth=2)
plt.plot(t, v, label='Kecepatan $v(t)$ (m/s)', color='blue', linewidth=2)
plt.plot(t, a, label='Percepatan $a(t)$ ($m/s^2$)', color='green', linewidth=2)

# Menambahkan elemen pelengkap informasi
plt.title('Grafik Kinematika Kendaraan: Posisi, Kecepatan, dan Percepatan terhadap Waktu', fontsize=14)
plt.xlabel('Waktu $t$ (detik)', fontsize=12)
plt.ylabel('Nilai Skala (m, m/s, $m/s^2$)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)

# Menampilkan grafik di layar
plt.show()
```
