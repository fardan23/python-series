```python
# buat file txt
nf = "r09.txt"
with open(nf, "w", encoding="utf-8") as f:
  f.write("Hello kelas RO \n")
  f.write("Belajar python tentang text \n")
  f.write("Selesai")
print(f"File {nf} created")

# baca file txt
with open(nf, "r", encoding="utf-8") as f:
  isi_teks = f.read()
  print("Isi file: ")
  print(isi_teks)
```

```python
# pengolahan
data = [
    "Fardan:100",
    "Ruka:95",
    "Pharita:90",
    "Asa:85",
    "Ahyeon:80",
    "Rami:75",
    "Rora:70",
    "Chiquita:65"
]
total_nilai = 0
jumlah_data = len(data)

for baris in data:
  baris_bersih = baris.strip()
  if ":" in baris_bersih:
    nama, nilai_str = baris_bersih.split(":")
    nilai = int(nilai_str)
    total_nilai += int(nilai)
    print(f"-> Siswa: {nama}, | Nilai: {nilai}")
rata_rata = total_nilai / jumlah_data
print(f"Rata-rata kelas: {rata_rata}")
```

```python
# buat file csv
import csv

nf_csv = "data_mhs.csv"
data_mhs = [
    ["NPM", "Nama", "Prodi"],
    [2025435001, "Fardan", "TI"],
    [2025435002, "Ruka", "SI"],
    [2025435003, "Pharita", "BK"],
    [2025435004, "Asa", "BD"],
    [2025435005, "Ahyeon", "MR"],
    [2025435006, "Rami", "SD"],
    [2025435007, "Rora", "PBI"],
    [2025435008, "Chiquita", "PM"]
]

# menulis csv
with open(nf_csv, "w", encoding="utf-8", newline="") as f:
  writer = csv.writer(f)
  writer.writerows(data_mhs)
print(f"File {nf_csv} created")

# membaca csv
with open(nf_csv, "r", encoding="utf-8") as f:
  reader = csv.reader(f)
  for baris in reader:
    print(baris)
print("\n Proses selesai")
```

```python
# json
import json
import os

print("Pengolahan json")
nf_json = "prodi.json"

data_prodi = {
    "aplikasi" : "Sistem Data Kampus",
    "versi" : "1.0.0",
    "aktif" : True,
    "prodi" : ["TI", "MI", "TK", "AK", "AR", "BD", "FIS", "KIM"]
}

# nulis json
with open(nf_json, "w", encoding="utf-8") as f:
  json.dump(data_prodi, f, indent=4, ensure_ascii=False)
print(f"File {nf_json} created")

# baca json
with open(nf_json, "r", encoding="utf-8") as f:
  data_json = json.load(f)
  print(data_json)
  print(f"Aplikasi: {data_json['aplikasi']}")
  print(f"Versi: {data_json['versi']}")
  print(f"Aktif: {data_json['aktif']}")
  print(f"Prodi: {', '.join(data_json['prodi'])}")
```
